```python
import pandas as pd
from pathlib import Path
import os
```

```python
BASE_PATH = Path("../data/raw")

print("Base path:", BASE_PATH.resolve())
```

```python
files = {
    "CLN1_Patient": BASE_PATH / "cln1" / "cln01_patient_table_jul25.xlsx",
    "CLN1_Mutation": BASE_PATH / "cln1" / "cln01_mutation_table_jul25.xlsx",

    "CLN2_Patient": BASE_PATH / "cln2" / "cln02_patient_table_jan26.xlsx",
    "CLN2_Mutation": BASE_PATH / "cln2" / "cln02_mutation_table_jan26.xlsx",

    "CLN3_Patient": BASE_PATH / "cln3" / "cln03_patient_table_feb26.xlsx",
    "CLN3_Mutation": BASE_PATH / "cln3" / "cln03_mutation_table_feb26.xlsx"
}

for name, path in files.items():
    print(name)
    print("Exists:", path.exists())
    print("Path:", path)
    print("-" * 60)
```

```python
for name, path in files.items():
    excel_file = pd.ExcelFile(path)

    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)
    print("Sheets:", excel_file.sheet_names)
```

```python
dataset_info = {}

for name, path in files.items():

    print("\n" + "=" * 80)
    print("DATASET:", name)
    print("=" * 80)

    # Read first sheet
    df_temp = pd.read_excel(path)

    # Store information
    dataset_info[name] = {
        "shape": df_temp.shape,
        "columns": df_temp.columns.tolist()
    }

    print("\nSHAPE:")
    print(df_temp.shape)

    print("\nCOLUMNS:")
    for i, col in enumerate(df_temp.columns):
        print(f"{i}: {repr(col)}")

    print("\nFIRST 10 ROWS:")
    display(df_temp.head(10))

    print("\nMISSING VALUES:")
    print(df_temp.isnull().sum())

    print("\nEXACT DUPLICATE ROWS:")
    print(df_temp.duplicated().sum())
    print("\n\n" + "=" * 80)
print("FINAL DATASET SUMMARY")
print("=" * 80)

for name, info in dataset_info.items():
    print("\nDATASET:", name)
    print("Shape:", info["shape"])
    print("Columns:")
    print(info["columns"])
```
