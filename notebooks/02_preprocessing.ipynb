```python
def find_header_row(file_path, sheet_name, keywords, max_rows=30):

    raw_df = pd.read_excel(
        file_path,
        sheet_name=sheet_name,
        header=None,
        nrows=max_rows
    )

    for index, row in raw_df.iterrows():

        row_text = " ".join(
            str(value).lower()
            for value in row.values
            if pd.notna(value)
        )

        matches = sum(
            keyword.lower() in row_text
            for keyword in keywords
        )

        if matches >= 2:
            return index

    return None
```

```python
def find_header_row(file_path, sheet_name=0, search_text="Patient"):
    
    temp = pd.read_excel(
        file_path,
        sheet_name=sheet_name,
        header=None
    )

    for index, row in temp.iterrows():

        # Convert every value safely into a string
        row_text = " ".join(
            str(value) for value in row.tolist()
        ).lower()

        if search_text.lower() in row_text:
            return index

    return None


# Find CLN3 Patient Header
cln3_patient_header = find_header_row(
    cln3_patient_path,
    search_text="Patient"
)


# Find CLN3 Mutation Header
cln3_mutation_header = find_header_row(
    cln3_mutation_path,
    search_text="Identifier"
)


print("CLN3 Patient Header:", cln3_patient_header)
print("CLN3 Mutation Header:", cln3_mutation_header)
```

```python
# ============================================================
# CLN3 MUTATION MATCHING ANALYSIS
# ============================================================

# Clean text values first
patient_mutations = (
    cln3_patient["mutation number"]
    .dropna()
    .astype(str)
    .str.strip()
)

mutation_identifiers = (
    cln3_mutation["Identifier"]
    .dropna()
    .astype(str)
    .str.strip()
)

patient_unique_mutations = set(patient_mutations.unique())
mutation_unique_identifiers = set(mutation_identifiers.unique())

matched_mutations = (
    patient_unique_mutations
    & mutation_unique_identifiers
)

unmatched_mutations = (
    patient_unique_mutations
    - mutation_unique_identifiers
)

print("=" * 60)
print("CLN3 MUTATION MATCHING")
print("=" * 60)

print("\nUnique mutations in Patient Dataset:")
print(len(patient_unique_mutations))

print("\nUnique identifiers in Mutation Dataset:")
print(len(mutation_unique_identifiers))

print("\nMatched mutations:")
print(len(matched_mutations))

print("\nUnmatched patient mutations:")
print(len(unmatched_mutations))

print("\nSample unmatched mutations:")
print(list(unmatched_mutations)[:20])
```

```python
# ============================================================
# CLEAN CLN3 PATIENT DATASET
# ============================================================

print("Before removing exact duplicates:", len(cln3_patient))

# Remove exact duplicate rows
cln3_patient_clean = cln3_patient.drop_duplicates().copy()

# Remove completely empty rows again
cln3_patient_clean = cln3_patient_clean.dropna(how="all")

# Clean Patient ID
cln3_patient_clean["DB Patient ID"] = (
    cln3_patient_clean["DB Patient ID"]
    .astype("string")
    .str.strip()
)

# Clean mutation number
cln3_patient_clean["mutation number"] = (
    cln3_patient_clean["mutation number"]
    .astype("string")
    .str.strip()
    .str.lower()
)

print("After removing exact duplicates:", len(cln3_patient_clean))
print("Exact duplicates remaining:", cln3_patient_clean.duplicated().sum())

print("\nUnique Patient IDs:",
      cln3_patient_clean["DB Patient ID"].nunique())

print("Missing Patient IDs:",
      cln3_patient_clean["DB Patient ID"].isna().sum())
```

```python
# ============================================================
# CLEAN CLN3 MUTATION DATASET
# ============================================================

print("Original mutation rows:", len(cln3_mutation))

# Remove completely empty rows
cln3_mutation_clean = cln3_mutation.dropna(how="all").copy()

# Clean Identifier
cln3_mutation_clean["Identifier"] = (
    cln3_mutation_clean["Identifier"]
    .astype("string")
    .str.strip()
    .str.lower()
)

# Remove rows where Identifier is missing
cln3_mutation_clean = cln3_mutation_clean.dropna(
    subset=["Identifier"]
)

# Remove duplicate mutation identifiers
# Keep first occurrence
cln3_mutation_clean = (
    cln3_mutation_clean
    .drop_duplicates(
        subset=["Identifier"],
        keep="first"
    )
)

print("Cleaned mutation rows:", len(cln3_mutation_clean))

print("Missing Identifiers:",
      cln3_mutation_clean["Identifier"].isna().sum())

print("Unique Identifiers:",
      cln3_mutation_clean["Identifier"].nunique())

print("Duplicate Identifiers:",
      cln3_mutation_clean["Identifier"].duplicated().sum())
```

```python
# ============================================================
# MERGE CLN3 PATIENT + MUTATION DATA
# ============================================================

cln3_merged = cln3_patient_clean.merge(

    cln3_mutation_clean,

    how="left",

    left_on="mutation number",

    right_on="Identifier",

    suffixes=(
        "_patient",
        "_mutation"
    )
)

print("=" * 60)
print("CLN3 MERGED DATASET")
print("=" * 60)

print("\nMerged dataset shape:")
print(cln3_merged.shape)

print("\nPatients with matched mutation information:")

matched_count = cln3_merged["Identifier"].notna().sum()

print(matched_count)

print("\nPatients WITHOUT matched mutation information:")

unmatched_count = cln3_merged["Identifier"].isna().sum()

print(unmatched_count)
```

```python
# ============================================================
# ADD DISEASE SUBTYPE
# ============================================================

cln3_merged["Disease_Subtype"] = "CLN3"

print(
    cln3_merged["Disease_Subtype"]
    .unique()
)
```

```python
# ============================================================
# SAVE CLN3 PROCESSED DATASET
# ============================================================

import os

processed_folder = "../data/processed"

os.makedirs(
    processed_folder,
    exist_ok=True
)

output_path = (
    "../data/processed/"
    "cln3_processed.csv"
)

cln3_merged.to_csv(
    output_path,
    index=False
)

print("CLN3 processed dataset saved successfully!")
print("Location:", output_path)

print("\nFinal Dataset Shape:")
print(cln3_merged.shape)
```
